#  Data Validation: Check Price & Discount Consistency

```python
#Calculate mathematically expected price based on MRP and discount percentage
expected_descounted_price=df['mrp']*(1-df['discountPercent']/100)
expected_descounted_price
```

```python
df['discountedSellingPrice']=df['mrp']*(1-df['discountPercent']/100)
```

```python
# Identify rows with calculation discrepancies greater than 1 unit due to rounding errors
price_discrepancies=df[abs(df['discountedSellingPrice']-expected_descounted_price)>1]
print(f"[ok]{len(price_discrepancies)}")
```

# featurre engineering

```python
# 1 Calculate actual monetary savings for the consumer
df['money_saved']=df['mrp']-df['discountedSellingPrice']
```

```python
# 3 Calculate total financial value of current available stock
df['total_stock_value']=df['availableQuantity']*df['discountedSellingPrice']
```

#  SQL Optimization: Data Type Conversion

```python
# Convert outOfStock boolean values to integers (1 for True, 0 for False)
df['outOfStock']=df['outOfStock'].astype(int)
```

```python
#  Preview Processed Dataset & Verify New Columns
print(df[['Category','name','money_saved','outOfStock']].head())
```
